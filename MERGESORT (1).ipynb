```python
%%writefile merge.c
#include <stdio.h>

int count = 0;

void merge(int a[], int l, int m, int r)
{
    int i = l, j = m + 1, k = 0;
    int temp[20];

    while (i <= m && j <= r)
    {
        count++;

        if (a[i] <= a[j])
            temp[k++] = a[i++];
        else
            temp[k++] = a[j++];
    }

    while (i <= m)
        temp[k++] = a[i++];

    while (j <= r)
        temp[k++] = a[j++];

    for (i = l, k = 0; i <= r; i++)
        a[i] = temp[k++];
}

void mergeSort(int a[], int l, int r)
{
    if (l < r)
    {
        int m = (l + r) / 2;

        mergeSort(a, l, m);
        mergeSort(a, m + 1, r);
        merge(a, l, m, r);
    }
}

int main()
{
    int a[] = {50, 20, 40, 10, 30};
    int n = 5;
    int i;

    printf("Original marks: ");

    for (i = 0; i < n; i++)
        printf("%d ", a[i]);

    mergeSort(a, 0, n - 1);

    printf("\nSorted marks in ascending order: ");

    for (i = 0; i < n; i++)
        printf("%d ", a[i]);

    printf("\nNumber of comparisons: %d\n", count);

    return 0;
}
```
